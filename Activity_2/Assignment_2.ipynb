# ACTIVITY 1: IMDB SENTIMENT CLASSIFICATION

```python
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)
import tiktoken

# Reproducibility
torch.manual_seed(42)
np.random.seed(42)

# Device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)
```

Load Dataset

```python
import pandas as pd

df = pd.read_csv(
    "IMDB Dataset.csv",
    encoding="utf-8",
    engine="python",
    on_bad_lines="skip"
)

print("Rows loaded:", len(df))
print("\nColumns:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum())

print("\nSentiment:")
print(df["sentiment"].value_counts())
```

```python
# Remove missing values and duplicate reviews

df = df.dropna(subset=["review", "sentiment"])
df = df.drop_duplicates(subset=["review"]).reset_index(drop=True)

print("Total usable reviews:", len(df))
print(df["sentiment"].value_counts())
```

```python
# Use 95% of the available dataset

df = df.sample(frac=0.95, random_state=42).reset_index(drop=True)

print("Reviews used (95%):", len(df))
```

Convert sentiment to numbers

```python
df["sentiment"] = df["sentiment"].map({
    "negative": 0,
    "positive": 1
})

print(df["sentiment"].value_counts())
```

Train/Test split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    df["review"].values,
    df["sentiment"].values,
    test_size=0.05,
    random_state=42,
    stratify=df["sentiment"].values
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

Import tiktoken

```python
import tiktoken

encoding = tiktoken.get_encoding("cl100k_base")

print("Tokenizer loaded successfully")
print("Vocabulary size:", encoding.n_vocab)
```

Tokenization + Padding

```python
MAX_LEN = 200
PAD_ID = 0

def tokenize_and_pad(texts):
    data = []

    for text in texts:
        tokens = encoding.encode(text)

        # Truncate
        tokens = tokens[:MAX_LEN]

        # Shift IDs by 1 because 0 is used for padding
        tokens = [token + 1 for token in tokens]

        # Pad
        tokens += [PAD_ID] * (MAX_LEN - len(tokens))

        data.append(tokens)

    return torch.tensor(data, dtype=torch.long)
```

Tokenize train and test data

```python
import torch

X_train_tensor = tokenize_and_pad(X_train)
X_test_tensor = tokenize_and_pad(X_test)

y_train_tensor = torch.tensor(y_train, dtype=torch.long)
y_test_tensor = torch.tensor(y_test, dtype=torch.long)

print("X_train:", X_train_tensor.shape)
print("X_test :", X_test_tensor.shape)
print("y_train:", y_train_tensor.shape)
print("y_test :", y_test_tensor.shape)
```

Create DataLoaders

```python
from torch.utils.data import TensorDataset, DataLoader

train_dataset = TensorDataset(
    X_train_tensor,
    y_train_tensor
)

test_dataset = TensorDataset(
    X_test_tensor,
    y_test_tensor
)

train_loader = DataLoader(
    train_dataset,
    batch_size=128,
    shuffle=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=128,
    shuffle=False
)

print("DataLoaders created successfully")
```

```python
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Using:", device)
```

Build the Improved RNN

```python
import torch.nn as nn

class RNNClassifier(nn.Module):

    def __init__(self, vocab_size):
        super().__init__()

        self.embedding = nn.Embedding(
            vocab_size,
            128,
            padding_idx=PAD_ID
        )

        self.rnn = nn.RNN(
            input_size=128,
            hidden_size=128,
            num_layers=2,
            batch_first=True,
            dropout=0.3,
            bidirectional=True
        )

        self.dropout = nn.Dropout(0.4)

        self.fc = nn.Linear(
            128 * 2,
            2
        )

    def forward(self, x):

        x = self.embedding(x)

        output, hidden = self.rnn(x)

        # Use final timestep
        x = output[:, -1, :]

        x = self.dropout(x)

        return self.fc(x)
```

Create Model

```python
VOCAB_SIZE = encoding.n_vocab + 1

model = RNNClassifier(
    VOCAB_SIZE
).to(device)

print(model)
```

Loss and Optimizer

```python
criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=0.001,
    weight_decay=1e-4
)

print("Loss and optimizer ready")
```

Train the RNN

```python
EPOCHS = 8

best_loss = float("inf")
patience = 2
counter = 0

for epoch in range(EPOCHS):

    model.train()

    total_loss = 0

    for batch_x, batch_y in train_loader:

        batch_x = batch_x.to(device)
        batch_y = batch_y.to(device)

        optimizer.zero_grad()

        output = model(batch_x)

        loss = criterion(output, batch_y)

        loss.backward()

        # Gradient clipping
        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=1.0
        )

        optimizer.step()

        total_loss += loss.item()

    avg_loss = total_loss / len(train_loader)

    print(
        f"Epoch {epoch+1}/{EPOCHS} "
        f"- Loss: {avg_loss:.4f}"
    )

    # Save best model
    if avg_loss < best_loss:

        best_loss = avg_loss
        counter = 0

        torch.save(
            model.state_dict(),
            "best_rnn_model.pth"
        )

    else:

        counter += 1

        if counter >= patience:

            print("Early stopping...")
            break
```

Load Best Model

```python
model.load_state_dict(
    torch.load(
        "best_rnn_model.pth",
        map_location=device
    )
)

model.eval()

print("Best RNN model loaded")
```

Make Test Predictions

```python
all_predictions = []
all_labels = []

model.eval()

with torch.no_grad():

    for batch_x, batch_y in test_loader:

        batch_x = batch_x.to(device)

        output = model(batch_x)

        predictions = torch.argmax(
            output,
            dim=1
        ).cpu().numpy()

        all_predictions.extend(predictions)
        all_labels.extend(batch_y.numpy())

print("Predictions completed")
```

Accuracy, Precision, Recall, F1

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

accuracy = accuracy_score(
    all_labels,
    all_predictions
)

precision = precision_score(
    all_labels,
    all_predictions
)

recall = recall_score(
    all_labels,
    all_predictions
)

f1 = f1_score(
    all_labels,
    all_predictions
)

print("========== RNN RESULTS ==========")
print(f"Accuracy : {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1 Score : {f1:.4f}")
```

Classification Report

```python
from sklearn.metrics import classification_report

print(
    classification_report(
        all_labels,
        all_predictions,
        target_names=[
            "Negative",
            "Positive"
        ]
    )
)
```

Confusion Matrix

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(
    all_labels,
    all_predictions
)

print("Confusion Matrix:")
print(cm)
```

Visualize Confusion Matrix

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 5))

plt.imshow(cm)

plt.title("RNN Confusion Matrix")
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")

plt.xticks(
    [0, 1],
    ["Negative", "Positive"]
)

plt.yticks(
    [0, 1],
    ["Negative", "Positive"]
)

for i in range(2):
    for j in range(2):
        plt.text(
            j,
            i,
            cm[i, j],
            ha="center",
            va="center"
        )

plt.colorbar()
plt.show()
```

# Activity 2 - Text Classification with LSTM

LSTM Model

```python
import torch
import torch.nn as nn

class LSTMClassifier(nn.Module):

    def __init__(self, vocab_size):
        super().__init__()

        # Embedding layer
        self.embedding = nn.Embedding(
            vocab_size,
            128,
            padding_idx=PAD_ID
        )

        # LSTM layer
        self.lstm = nn.LSTM(
            input_size=128,
            hidden_size=128,
            num_layers=2,
            batch_first=True,
            dropout=0.3,
            bidirectional=True
        )

        # Dropout
        self.dropout = nn.Dropout(0.4)

        # Linear output layer
        self.fc = nn.Linear(
            128 * 2,
            2
        )

    def forward(self, x):

        # Embedding
        x = self.embedding(x)

        # LSTM
        output, (hidden, cell) = self.lstm(x)

        # Last timestep
        x = output[:, -1, :]

        # Dropout
        x = self.dropout(x)

        # Classification
        x = self.fc(x)

        return x
```

Create the LSTM Model

```python
VOCAB_SIZE = encoding.n_vocab + 1

lstm_model = LSTMClassifier(
    VOCAB_SIZE
).to(device)

print(lstm_model)
```

Loss Function and Optimizer

```python
criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.AdamW(
    lstm_model.parameters(),
    lr=0.001,
    weight_decay=1e-4
)
```

Train the LSTM

```python
EPOCHS = 8

best_loss = float("inf")
patience = 2
counter = 0

for epoch in range(EPOCHS):

    lstm_model.train()

    total_loss = 0

    for batch_x, batch_y in train_loader:

        batch_x = batch_x.to(device)
        batch_y = batch_y.to(device)

        # Clear gradients
        optimizer.zero_grad()

        # Forward pass
        output = lstm_model(batch_x)

        # Calculate loss
        loss = criterion(output, batch_y)

        # Backpropagation
        loss.backward()

        # Prevent exploding gradients
        torch.nn.utils.clip_grad_norm_(
            lstm_model.parameters(),
            max_norm=1.0
        )

        # Update weights
        optimizer.step()

        total_loss += loss.item()

    avg_loss = total_loss / len(train_loader)

    print(
        f"Epoch {epoch + 1}/{EPOCHS} "
        f"- Training Loss: {avg_loss:.4f}"
    )

    # Save best model
    if avg_loss < best_loss:

        best_loss = avg_loss
        counter = 0

        torch.save(
            lstm_model.state_dict(),
            "best_lstm_model.pth"
        )

    else:

        counter += 1

        if counter >= patience:

            print("Early stopping...")
            break
```

Load Best LSTM Model

```python
lstm_model.load_state_dict(
    torch.load(
        "best_lstm_model.pth",
        map_location=device
    )
)

lstm_model.eval()

print("Best LSTM model loaded successfully.")
```

Make Predictions on Test Set

```python
lstm_predictions = []
lstm_labels = []

lstm_model.eval()

with torch.no_grad():

    for batch_x, batch_y in test_loader:

        batch_x = batch_x.to(device)

        output = lstm_model(batch_x)

        predictions = torch.argmax(
            output,
            dim=1
        ).cpu().numpy()

        lstm_predictions.extend(predictions)
        lstm_labels.extend(batch_y.numpy())

print("LSTM predictions completed.")
```

Accuracy, Precision, Recall and F1

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

lstm_accuracy = accuracy_score(
    lstm_labels,
    lstm_predictions
)

lstm_precision = precision_score(
    lstm_labels,
    lstm_predictions
)

lstm_recall = recall_score(
    lstm_labels,
    lstm_predictions
)

lstm_f1 = f1_score(
    lstm_labels,
    lstm_predictions
)

print("========== LSTM RESULTS ==========")
print(f"Accuracy : {lstm_accuracy:.4f}")
print(f"Precision: {lstm_precision:.4f}")
print(f"Recall   : {lstm_recall:.4f}")
print(f"F1 Score : {lstm_f1:.4f}")
```

Classification Report

```python
from sklearn.metrics import classification_report

print(
    classification_report(
        lstm_labels,
        lstm_predictions,
        target_names=[
            "Negative",
            "Positive"
        ]
    )
)
```

Confusion Matrix

```python
from sklearn.metrics import confusion_matrix

lstm_cm = confusion_matrix(
    lstm_labels,
    lstm_predictions
)

print("LSTM Confusion Matrix:")
print(lstm_cm)
```

### 4. Based on this exercise, what is the LSTM doing differently from the RNN that could explain any difference in results?

The LSTM is designed to handle long-term dependencies in sequential data more effectively than a standard RNN. It uses a memory cell and gating mechanisms to control what information should be **retained, updated, and forgotten** while processing the sequence.

A standard RNN can have difficulty retaining important information from earlier parts of a long review because of problems such as vanishing gradients. This can make it harder to understand sentiment when important words or phrases occur far apart in the review.

The LSTM's gating mechanism helps preserve relevant information for longer sequences and discard information that is less useful. This can explain why the LSTM achieved much better results in this experiment.

In our experiment, this difference is clearly reflected in the results:

- RNN Accuracy = **62.22%**
- LSTM Accuracy = **86.76%**
- RNN Training Loss = **0.6225**
- LSTM Training Loss = **0.0813**

Thus, the LSTM was able to learn the sentiment patterns in the movie reviews much more effectively than the RNN.
