```python
import pandas as pd
import torch
import torch.nn as nn
import tiktoken
from torch.utils.data import TensorDataset, DataLoader
from sklearn.model_selection import train_test_split

torch.manual_seed(1)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using:", device)
```

```python
df = pd.read_csv("IMDB Dataset.csv")

print("Dataset shape:", df.shape)
print(df.head())
print(df.columns)

print("\nFull dataset loaded successfully!")
```

```python
# 2. Convert sentiment into numeric labels
# positive = 1, negative = 0

df["sentiment"] = df["sentiment"].map({
    "positive": 1,
    "negative": 0
})

texts = df["review"].tolist()
labels = df["sentiment"].tolist()

print("Number of reviews:", len(texts))
print("Number of labels:", len(labels))
print("Label counts:")
print(df["sentiment"].value_counts())
```

```python
# 3. Split the data into training and test sets

train_texts, test_texts, train_labels, test_labels = train_test_split(
    texts,
    labels,
    test_size=0.20,
    random_state=1,
    stratify=labels
)

print("Training samples:", len(train_texts))
print("Testing samples:", len(test_texts))
```

```python
# 4. Tokenize using tiktoken (cl100k_base)

enc = tiktoken.get_encoding("cl100k_base")

print("Tokenizer:", "cl100k_base")
print("Tokenizer vocabulary size:", enc.n_vocab)

sample_tokens = enc.encode(train_texts[0])

print("\nSample review:")
print(train_texts[0][:300])
print("\nFirst 20 token IDs:")
print(sample_tokens[:20])
```

```python
# 5. Pad or truncate token sequences to a fixed length

MAX_LENGTH = 200

# tiktoken has no dedicated padding token.
# We create one outside the tokenizer's normal token-ID range.
PAD_ID = enc.n_vocab

def encode_and_pad(text):
    tokens = enc.encode(text)

    # Truncate long reviews
    tokens = tokens[:MAX_LENGTH]

    # Pad short reviews
    if len(tokens) < MAX_LENGTH:
        tokens = tokens + [PAD_ID] * (MAX_LENGTH - len(tokens))

    return tokens

encoded_sample = encode_and_pad(train_texts[0])

print("Encoded review length:", len(encoded_sample))
print("First 30 token IDs:")
print(encoded_sample[:30])
```

```python
# 6. Convert token sequences and labels into PyTorch tensors

x_train = [encode_and_pad(text) for text in train_texts]
x_test = [encode_and_pad(text) for text in test_texts]

X_train = torch.tensor(x_train, dtype=torch.long)
y_train = torch.tensor(train_labels, dtype=torch.float32)

X_test = torch.tensor(x_test, dtype=torch.long)
y_test = torch.tensor(test_labels, dtype=torch.float32)

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)
print("X_test :", X_test.shape)
print("y_test :", y_test.shape)
```

```python
# Create DataLoaders

BATCH_SIZE = 64

train_dataset = TensorDataset(X_train, y_train)
test_dataset = TensorDataset(X_test, y_test)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False
)

print("Training batches:", len(train_loader))
print("Testing batches:", len(test_loader))
```

```python
import torch.nn as nn

class MovieRNN(nn.Module):
    def __init__(self, vocab_size, embedding_dim=128, hidden_size=128, padding_idx=None, dropout_prob=0.5):
        super().__init__()

        self.embedding = nn.Embedding(
            vocab_size,
            embedding_dim,
            padding_idx=padding_idx
        )

        # Changed from nn.RNN to nn.LSTM
        self.rnn = nn.LSTM(
            embedding_dim,
            hidden_size,
            batch_first=True
        )

        self.dropout = nn.Dropout(dropout_prob)

        self.output = nn.Linear(
            hidden_size,
            1
        )

    def forward(self, x):
        x = self.embedding(x)

        # LSTM returns (output, (h_n, c_n)), we need h_n for the last hidden state
        _, (hidden_n, _) = self.rnn(x)

        # Apply dropout to the last hidden state before the linear layer
        # hidden_n has shape (num_layers * num_directions, batch_size, hidden_size)
        # For a single layer, single direction LSTM, we take the first (and only) hidden state
        hidden = self.dropout(hidden_n[-1])

        return self.output(hidden).squeeze(1)


# +1 because PAD_ID is outside the tokenizer vocabulary
movie = MovieRNN(
    vocab_size=enc.n_vocab + 1,
    embedding_dim=128, # Increased embedding_dim
    hidden_size=128,   # Increased hidden_size
    padding_idx=PAD_ID,
    dropout_prob=0.5
).to(device)

print(movie)
```

```python
# 8. Loss function and optimizer

loss_fn = nn.BCEWithLogitsLoss()

optimizer = torch.optim.Adam(
    movie.parameters(),
    lr=0.001
)

print("Loss:", loss_fn)
print("Optimizer:", optimizer.__class__.__name__)
```

```python
# Train the RNN

EPOCHS = 10 # Increased number of epochs

for epoch in range(EPOCHS):

    # Training Phase
    movie.train()

    train_loss = 0.0
    train_correct = 0
    train_total = 0

    for X_batch, y_batch in train_loader:

        X_batch = X_batch.to(device)
        y_batch = y_batch.to(device)

        # Forward pass
        logits = movie(X_batch)

        # Calculate loss
        loss = loss_fn(logits, y_batch)

        # Clear old gradients
        optimizer.zero_grad()

        # Backpropagation
        loss.backward()

        # Update weights
        optimizer.step()

        train_loss += loss.item()

        # Convert logits to probabilities
        probabilities = torch.sigmoid(logits)

        predictions = (probabilities >= 0.5).float()

        train_correct += (predictions == y_batch).sum().item()
        train_total += y_batch.size(0)

    avg_train_loss = train_loss / len(train_loader)
    train_accuracy = train_correct / train_total

    # Evaluation Phase
    movie.eval()

    test_loss = 0.0
    test_correct = 0
    test_total = 0

    with torch.no_grad():
        for X_batch, y_batch in test_loader:
            X_batch = X_batch.to(device)
            y_batch = y_batch.to(device)

            logits = movie(X_batch)
            loss = loss_fn(logits, y_batch)
            test_loss += loss.item()

            probabilities = torch.sigmoid(logits)
            predictions = (probabilities >= 0.5).float()

            test_correct += (predictions == y_batch).sum().item()
            test_total += y_batch.size(0)

    avg_test_loss = test_loss / len(test_loader)
    test_accuracy = test_correct / test_total

    print(
        f"Epoch {epoch + 1}/{EPOCHS} | "
        f"Train Loss = {avg_train_loss:.4f} | "
        f"Train Accuracy = {train_accuracy:.4f} | "
        f"Test Loss = {avg_test_loss:.4f} | "
        f"Test Accuracy = {test_accuracy:.4f}"
    )
```

```python
# The evaluation is now part of the training loop.
# This cell can be removed or kept for a final evaluation after training.
# For now, I will keep it as a redundant check, but it's not strictly necessary with the updated training loop.

movie.eval()

correct = 0
total = 0
test_loss_final = 0.0

with torch.no_grad():

    for X_batch, y_batch in test_loader:

        X_batch = X_batch.to(device)
        y_batch = y_batch.to(device)

        logits = movie(X_batch)

        loss = loss_fn(logits, y_batch)
        test_loss_final += loss.item()

        probabilities = torch.sigmoid(logits)
        predictions = (probabilities >= 0.5).float()

        correct += (predictions == y_batch).sum().item()
        total += y_batch.size(0)

test_accuracy_final = correct / total
average_test_loss_final = test_loss_final / len(test_loader)

print(f"Final Test Loss: {average_test_loss_final:.4f}")
print(f"Final Test Accuracy: {test_accuracy_final:.4f}")
```

```python
# Predict the sentiment of a new review

def predict_sentiment(text):
    movie.eval()

    encoded = encode_and_pad(text)

    x = torch.tensor(
        [encoded],
        dtype=torch.long
    ).to(device)

    with torch.no_grad():
        logit = movie(x)
        probability = torch.sigmoid(logit).item()

    if probability >= 0.5:
        sentiment = "Positive"
        confidence = probability
    else:
        sentiment = "Negative"
        confidence = 1 - probability

    print("Sentiment:", sentiment)
    print(f"Confidence: {confidence:.4f}")


predict_sentiment(
    "This movie was absolutely fantastic and I really enjoyed it."
)
```
